```python
import sys
import onnx
import pathlib

block = "mha"
onnx_model=f"transformer_{block}.onnx"
print("Python:", sys.executable)
print("ONNX version:", onnx.__version__)
print("ONNX path:", onnx.__file__)
print("Checker IR support:", onnx.IR_VERSION)

input_path = pathlib.Path(onnx_model)
onnx_model = onnx.load(input_path)

print("Model IR:", onnx_model.ir_version)
print(
    "Model opsets:",
    [(op.domain or "ai.onnx", op.version)
     for op in onnx_model.opset_import]
)

onnx.checker.check_model(onnx_model)
```
